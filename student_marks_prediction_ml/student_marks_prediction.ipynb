```python
# Import libraries

from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create dataset

# Feature: Study Hours
study_hours = np.array([[1], [2], [3], [4], [5], [6], [7], [8], [2.5], [3.5]])

# Target: Marks
marks = np.array([30, 40, 50, 60, 70, 80, 90, 95, 35, 55])
```

```python
# Create Linear Regression model

model = LinearRegression()
```

```python
# Train the model

model.fit(study_hours, marks)
```

```python
# Predict marks for new study hours

new_hours = [[6]]
prediction = model.predict(new_hours)
```

```python
# Print prediction result

print("Predicted marks for 6 study hours:", prediction[0])
```

```python
# Print model details

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
print("Model Accuracy:", model.score(study_hours, marks))
```

```python
# Visualize actual data and regression line

plt.figure(figsize=(8, 5))
plt.scatter(study_hours, marks, label="Actual Data")
plt.plot(study_hours, model.predict(study_hours), label="Regression Line")

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks Prediction")
plt.legend()
plt.grid(True)

# Save graph image
plt.savefig("student_marks_prediction_output.png")
```
